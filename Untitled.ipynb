```python
import pandas as pd

df = pd.read_excel("Stock_Detail.xls", header=8)
product_value = (
    df.groupby("NameToDisplay")["Stock Value"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

total_value = product_value["Stock Value"].sum()

product_value["cum_pct"] = (
    product_value["Stock Value"].cumsum()
    / total_value
) * 100

print(
    (product_value["cum_pct"] <= 80).sum()
)
```
